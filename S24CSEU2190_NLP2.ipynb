```python
import nltk
from nltk.corpus import wordnet

nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
words = [
    "car",
    "bank",
    "bat",
    "dog",
    "light",
    "plant",
    "book",
    "teacher",
    "student",
    "computer"
]
```

```python
for word in words:
    print("\nWord:", word)

    synsets = wordnet.synsets(word)

    for syn in synsets[:3]:
        print("Synset:", syn.name())
        print("Definition:", syn.definition())

        if syn.examples():
            print("Example:", syn.examples()[0])
        else:
            print("Example: No example available")
```

```python
print("Words with multiple meanings:\n")

for word in words:
    synsets = wordnet.synsets(word)

    if len(synsets) > 1:
        print(word, "has", len(synsets), "meanings")
```

```python
for word in ["bank", "bat", "light"]:
    print("\n", word.upper())

    for syn in wordnet.synsets(word)[:5]:
        print(syn.name(), ":", syn.definition())
```

```python
import pandas as pd
from nltk.corpus import wordnet
```

```python
words = [
    "dog",
    "cat",
    "car",
    "apple",
    "teacher",
    "student",
    "computer",
    "bird",
    "rose",
    "school"
]
```

```python
data = []

for word in words:

    synsets = wordnet.synsets(word, pos=wordnet.NOUN)

    if synsets:
        syn = synsets[0]
        hypernyms = syn.hypernyms()

        if hypernyms:
            hypernym = hypernyms[0].name()
        else:
            hypernym = "None"

        data.append([word, syn.name(), hypernym])

df = pd.DataFrame(
    data,
    columns=["Word", "Synset", "Direct Hypernym"]
)

df
```

```python
words = ["dog", "cat", "car", "bird", "rose"]
```

```python
for word in words:

    synsets = wordnet.synsets(word, pos=wordnet.NOUN)

    if synsets:
        syn = synsets[0]

        print("\n", word)

        for i in range(3):
            print("Level", i, ":", syn.name())

            hypernyms = syn.hypernyms()

            if not hypernyms:
                break

            syn = hypernyms[0]
```

```python
pairs = [
    ("car", "automobile"),
    ("dog", "cat"),
    ("dog", "animal"),
    ("dog", "computer"),
    ("car", "bus"),
    ("teacher", "student"),
    ("apple", "orange"),
    ("computer", "banana")
]
```

```python
results = []

for word1, word2 in pairs:

    syn1 = wordnet.synsets(word1, pos=wordnet.NOUN)[0]
    syn2 = wordnet.synsets(word2, pos=wordnet.NOUN)[0]

    similarity = syn1.path_similarity(syn2)

    results.append([
        word1,
        word2,
        syn1.name(),
        syn2.name(),
        similarity
    ])

df = pd.DataFrame(
    results,
    columns=[
        "Word 1",
        "Word 2",
        "Synset 1",
        "Synset 2",
        "Similarity"
    ]
)

df
```

```python
df = df.sort_values("Similarity", ascending=False)

df
```

```python
print("Observation:")
print("Words with similar meanings have higher similarity scores.")
print("Words with unrelated meanings have lower similarity scores.")
```
